```python
# 13.1

from datetime import date

now = date.today()
today_string = str(now)

with open('today.txt', 'w') as file:
  file.write(today_string)
```

```python
# 13.2

with open('today.txt', 'r') as file:
  today_string = file.read()

print("Today's date:", today_string)
```

```python
# 13.3

from datetime import datetime

format_str = '%Y-%m-%d'
today_date = datetime.strptime(today_string.strip(), format_str).date()

print("Parsed date:", today_date)
```

```python
# 15.1

import multiprocessing
import time
import random
from datetime import datetime

def print_time(process_id):
  wait_time = random.random()
  time.sleep(wait_time)

  current_time = datetime.now().strftime("%H:%M:%S.%f")
  print(f"Process {process_id}: waited {wait_time:.4f}s, time is {current_time}")

if __name__ == '__main__':
  processes = []

  for i in range(1, 4):
    p = multiprocessing.Process(target=print_time, args=(i,))
    processes.append(p)
    p.start()

  for p in processes:
    p.join()
```
